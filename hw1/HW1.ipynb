# Applying k-anonymity

```python
# Drop the identity
df_anon = df.drop(['PassengerId', 'Name'], axis=1)
df_anon.head()
```

```python
df_anon = df

# Parameter for the mondrian algorithm
qis = ['HomePlanet', 'CryoSleep', 'Destination', 'Age']
k =2
```

```python
def summarized(partition):
  for i in qis:
    # Integer formatting, and make sure the range > 0
    if pd.api.types.is_numeric_dtype(partition[i]) and partition[i].min() != partition[i].max():
      s = f"[{partition[i].min()}-{partition[i].max()}]"
    # Strings features formatting
    else:
      s = "{" + ",".join(map(str, partition[i].unique())) + "}"
    partition[i] = s
  return partition

def anonymize(partition, ranks):
  dim=ranks[0][0] # Use the highest variance feature to split

  # Split by median value for int, for more balanced data
  partition=partition.sort_values(by=dim)
  if pd.api.types.is_numeric_dtype(partition[dim]):
    mid = partition[dim].median()
    lhs = partition[partition[dim] <= mid]
    rhs = partition[partition[dim] > mid]
  # Split by position for non int values
  else:
    mid = len(partition) // 2
    lhs = partition.iloc[:mid]
    rhs = partition.iloc[mid:]

  # Recurse
  if(len(lhs)>=k and len(rhs)>=k):
    return pd.concat([mondrian(lhs), mondrian(rhs)])
  else:
    return summarized(partition)

def mondrian(partition):
  ranks={}

  for i in qis:
    # Count the Quasi identifier with their variance
    ranks[i] = partition[i].nunique()

  # Sort the ranks
  ranks = sorted(ranks.items(), key=lambda x: x[1], reverse=True)
  return anonymize(partition, ranks)
```

```python
df_anon = mondrian(df_anon)
```

```python
df_anon.head()
```

```python
small = df_anon.groupby(qis).size().min()
if small >= k:
  print(f"Safe since the smallest group size is {small} (less than k)")
else:
  print("Not Safe")
```

# ML with K-anonymity

```python
# Seperate the label from the dataset
X = df_anon.drop(columns=["Transported"])
y = df_anon["Transported"]
```

```python
# For interval dataset, change it to mid
def interval_to_mid(feat):
  if(feat, str) and '[' in feat:
    a,b = feat.strip('[]').split('-')
    return (int(a) + int(b)) // 2
  elif(feat, str) and '{' in feat:
    a = feat.strip('{}')
    return a
```

```python
def interval_to_mid(feat):
  if isinstance(feat, str) and '[' in feat:
    a,b = feat.strip('[]').split('-')
    # Convert to float first, then to int to handle '0.0', '1.0'
    return (int(float(a)) + int(float(b))) // 2
  elif isinstance(feat, str) and '{' in feat:
    a = feat.strip('{}')
    return a
  # Fallback in case a value doesn't match expected string formats
  return feat

X["Age"] = X["Age"].apply(interval_to_mid)
# X["CryoSleep"] = X["CryoSleep"].apply(interval_to_mid)
```

```python
# Change HomePlanet to list
def split_set(val):
    if isinstance(val, str) and "{" in val:
        return val.strip("{}").split(",")
    return [val]

X["HomePlanet"] = X["HomePlanet"].apply(split_set)
X["Destination"] = X["Destination"].apply(split_set)
```

```python
# Do multi hot encoding
from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()
X = X.join(pd.DataFrame(mlb.fit_transform(X.pop('HomePlanet')),
                          columns=mlb.classes_,
                          index=X.index))
X = X.join(pd.DataFrame(mlb.fit_transform(X.pop('Destination')),
                          columns=mlb.classes_,
                          index=X.index))
```

```python
X.head()
```

```python
X= pd.get_dummies(X)
```

```python
X.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size = 0.2, random_state = 42)
```

```python
model = SVC(kernel='rbf', C=1.0, gamma='scale')
model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

```python
print(f"The accuracy score is {accuracy_score(y_test, predictions)}")
print(f"The precision score is {precision_score(y_test, predictions)}")
print(f"The recall score is {recall_score(y_test, predictions)}")
print(f"The auc score is {roc_auc_score(y_test, predictions)}")
```
